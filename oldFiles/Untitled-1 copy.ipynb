```python
from string import Template
import yaml
import json

bundle_target = "prd"

# Lê o YAML como texto puro
with open("tabela1.yml", "r", encoding="utf-8") as f:
    yaml_text = f.read()

# Substitui variáveis no estilo ${bundle.target}
yaml_text = Template(yaml_text).safe_substitute({"{bundle.target}": bundle_target})

# Carrega YAML já resolvido
dados = yaml.safe_load(yaml_text)

tabelas = dados.get("tables", {})

for tabela_nome, tabela_info in tabelas.items():
    print(f"\n📌 Nome da tabela: {tabela_info.get('name', tabela_nome)}")

    # Metadata da tabela
    tabela_metadata = tabela_info.get("metadata", {})
    print("   Metadata:", json.dumps(tabela_metadata, indent=4, ensure_ascii=False) if tabela_metadata else "{}")

    colunas = tabela_info.get("columns", [])
    if colunas:
        print("   Colunas:")
        for coluna in colunas:
            nome_coluna = coluna.get("name", "sem_nome")
            tipo_coluna = coluna.get("type", "sem tipo")
            
            # Se tiver type -> adiciona 'certo'
            if "type" in coluna:
                print(f"     - {nome_coluna} ({tipo_coluna}) certo")
            else:
                print(f"     - {nome_coluna} ({tipo_coluna})")
            
            # Metadata da coluna
            coluna_metadata = coluna.get("metadata", {})
            print("         metadata:", json.dumps(coluna_metadata, indent=8, ensure_ascii=False) if coluna_metadata else "{}")
    else:
        print("   Nenhuma coluna")
```
